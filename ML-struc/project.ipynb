```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np 

# Load dataset
data = pd.read_csv("student_data.csv")

# Input/Output features selection based on your info.txt columns
X = data[['studytime']]
y = data['G3']

# Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predict scores for existing data
predicted_score = model.predict(X)

# Valid regression metrics calculation
mae = mean_absolute_error(y, predicted_score)
mse = mean_squared_error(y, predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted_score)

print("Mean Absolute Error:", round(mae, 2))
print("Mean Squared Error:", round(mse, 2))
print("Root Mean Squared Error:", round(rmse, 2))
print("R2 score:", round(r2, 4))

# Histogram Plot for Final Grade (G3)
plt.figure(figsize=(10, 6))
plt.hist(data["G3"], bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.grid(True)
plt.show()

# Scatter Plot with Regression Line 
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Actual Grades')
plt.plot(X, predicted_score, color='red', label='Predicted Grades (Regression Line)')
plt.title("Study Time vs Final Grade (G3)")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.legend()
plt.grid(True)
plt.show()

# Predicting for a new input value (e.g., studytime = 2)
new_study_time = 2
predicted_grade_new = model.predict([[new_study_time]])
print(f"Predicted final grade for study time {new_study_time} is: {round(predicted_grade_new[0], 2)}")
```
